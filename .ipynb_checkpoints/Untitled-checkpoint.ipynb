```python
"""
Quick-and-dirty modular helpers for analysing energy/accuracy sweep results.
Import in Jupyter and call analyze() per dataset, or use the pieces directly.

Assumes ONE experiment per file, with a single row where variant_id == 'baseline'.
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor

# columns that are always present regardless of data source
KNOWN_COLS = [
    'experiment_id', 'variant_id', 'rate', 'flops_reduced',
    'func_reduction_density_mean', 'func_reduction_density_std',
    'uc_depth_mean', 'uc_depth_std', 'uc_modularity',
    'energy_package_mean', 'energy_package_std',
    'energy_dram_mean', 'energy_dram_std',
    'energy_total_mean', 'energy_total_std',
]

# the "knobs" we want to test the influence of
FEATURE_COLS = [
    'rate', 'flops_reduced', 'func_reduction_density_mean',
    'func_reduction_density_std', 'uc_depth_mean', 'uc_depth_std',
    'uc_modularity',
]


# ---------------------------------------------------------------------------
# loading / prep
# ---------------------------------------------------------------------------

def load_data(path):
    """Read a results csv."""
    return pd.read_csv(path)


def get_result_cols(df):
    """Whatever columns aren't in KNOWN_COLS are treated as output/accuracy metrics."""
    return [c for c in df.columns if c not in KNOWN_COLS]


def add_baseline_diffs(df, result_cols=None):
    """Add a '<col>_diff' column = value - baseline value, for each result col.
    Assumes a single baseline row (variant_id == 'baseline') for this experiment."""
    df = df.copy()
    result_cols = result_cols or get_result_cols(df)

    baseline = df[df['variant_id'] == 'baseline']
    if baseline.empty:
        raise ValueError("No row with variant_id == 'baseline' found.")
    baseline = baseline.iloc[0]

    for col in result_cols:
        df[col + '_diff'] = df[col] - baseline[col]
    return df


def drop_baseline_row(df):
    """Remove the baseline row itself so it doesn't get analysed as a data point."""
    return df[df['variant_id'] != 'baseline'].copy()


def fill_failures_as_worst(df, result_cols, verbose=True):
    """Missing/NaN result = complete failure = worse than the worst observed value."""
    df = df.copy()
    for col in result_cols:
        dcol = col + '_diff'
        worst = df[dcol].min(skipna=True) - 3 * abs(df[dcol].std(skipna=True)) - 1
        n_missing = df[dcol].isna().sum()
        if n_missing and verbose:
            print(f"  {col}: {n_missing} missing/failed rows -> forced to worst diff ({worst:.3f})")
        df[dcol] = df[dcol].fillna(worst)
    return df


def get_targets(result_cols):
    """What we're trying to explain: total energy, plus each result's diff-to-baseline."""
    return ['energy_total_mean'] + [c + '_diff' for c in result_cols]


# ---------------------------------------------------------------------------
# influence / importance
# ---------------------------------------------------------------------------

def compute_correlations(df, feature_cols=FEATURE_COLS, targets=None):
    """Simple linear correlation of each feature against each target."""
    targets = targets or get_targets(get_result_cols(df))
    return df[feature_cols + targets].corr()[targets].loc[feature_cols].round(3)


def compute_rf_importance(df, feature_cols=FEATURE_COLS, targets=None, n_estimators=300):
    """Random forest feature importance per target (catches non-linear effects too)."""
    targets = targets or get_targets(get_result_cols(df))
    importances = {}
    for t in targets:
        sub = df.dropna(subset=[t])
        X = sub[feature_cols].fillna(0)
        y = sub[t]
        if y.nunique() <= 1 or len(y) < 3:
            continue
        rf = RandomForestRegressor(n_estimators=n_estimators, random_state=0)
        rf.fit(X, y)
        importances[t] = pd.Series(rf.feature_importances_, index=feature_cols)
    return pd.DataFrame(importances).round(3)


# ---------------------------------------------------------------------------
# plots (each returns a matplotlib Figure -> just call plt.show() in Jupyter,
# or `fig.savefig(path)` to save it)
# ---------------------------------------------------------------------------

def plot_importance(imp_df):
    """Bar chart of feature importance, one panel per target."""
    n = len(imp_df.columns)
    fig, axes = plt.subplots(1, n, figsize=(5 * n, 4), squeeze=False)
    for i, t in enumerate(imp_df.columns):
        imp_df[t].sort_values().plot.barh(ax=axes[0][i])
        axes[0][i].set_title(f"Influence on {t}")
    plt.tight_layout()
    return fig


def plot_tradeoff(df, result_col, color_by='rate'):
    """Scatter of energy_total_mean vs a result's diff-to-baseline (the accuracy/energy tradeoff)."""
    fig, ax = plt.subplots(figsize=(6, 5))
    sc = ax.scatter(df['energy_total_mean'], df[result_col + '_diff'],
                     c=df[color_by], cmap='viridis')
    fig.colorbar(sc, ax=ax, label=color_by)
    ax.set_xlabel('energy_total_mean')
    ax.set_ylabel(f'{result_col} diff vs baseline')
    ax.set_title(f'Energy vs accuracy tradeoff ({result_col})')
    plt.tight_layout()
    return fig


def plot_feature_vs_energy(df, feature_cols=FEATURE_COLS):
    """Scatter of each feature against energy_total_mean, side by side."""
    fig, axes = plt.subplots(1, len(feature_cols), figsize=(4 * len(feature_cols), 4))
    for ax, f in zip(axes, feature_cols):
        ax.scatter(df[f], df['energy_total_mean'], alpha=0.6)
        ax.set_xlabel(f)
        ax.set_ylabel('energy_total_mean')
    plt.tight_layout()
    return fig


# ---------------------------------------------------------------------------
# orchestrator - the one-call version for quick use
# ---------------------------------------------------------------------------

def analyze(path_or_df, feature_cols=FEATURE_COLS, verbose=True):
    """Run the full pipeline on one dataset (one experiment, one baseline row).

    Returns a dict:
        df          - working dataframe (baseline dropped, diff cols added, failures filled)
        result_cols - detected output/accuracy columns
        targets     - ['energy_total_mean', '<result>_diff', ...]
        corr        - correlation table (feature x target)
        importance  - random forest importance table (feature x target)
    """
    df = load_data(path_or_df) if isinstance(path_or_df, str) else path_or_df.copy()

    result_cols = get_result_cols(df)
    if verbose:
        print("Detected output/accuracy columns:", result_cols)

    df = add_baseline_diffs(df, result_cols)
    df = drop_baseline_row(df)
    df = fill_failures_as_worst(df, result_cols, verbose=verbose)

    targets = get_targets(result_cols)
    corr = compute_correlations(df, feature_cols, targets)
    importance = compute_rf_importance(df, feature_cols, targets)

    return {
        'df': df,
        'result_cols': result_cols,
        'targets': targets,
        'corr': corr,
        'importance': importance,
    }


def analyze_many(paths, feature_cols=FEATURE_COLS, verbose=False):
    """Run analyze() over multiple dataset files/dfs. Returns {name: result_dict}."""
    out = {}
    for p in paths:
        name = p if isinstance(p, str) else f"dataset_{len(out)}"
        if verbose:
            print(f"\n--- {name} ---")
        out[name] = analyze(p, feature_cols=feature_cols, verbose=verbose)
    return out
```
